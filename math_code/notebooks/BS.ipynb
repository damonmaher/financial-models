# Black-Scholes for Options Pricing

## Introduction

Options contracts are one of the most lucrative means of profit they are effectively excercised. So, there are several integral mathematical frameworks in the world of options pricing, which serve as the basis for financial models which optimize payouts for options contracts. Here, we construct one of the more well-known models: an implied volatility surface derived from the Black-Scholes framework. It's specifically called "Heston Stochastic Volatility". 

## The Options Problem

The Black-Scholes framework specifically refers to a system of partial differential equations and consequent solutions, which establish some notable axioms for options contracts. We specifically look at European call options, which give the buyer the choice, not the requirement, to purchase an asset at a strike price on the expiration date of the contract on top of an initial premium payment when the contract is initially purchased. But the risk here, is that the buyer does not know the value of the asset at this future expiration date. Solving for this is what options models are based around:
$$\operatorname{Payoff at time} T=\max(S_T-K, 0)$$
The profit is either the difference between the asset price at time $T$ and the strike price $K$. If the strike price ended up being more than the asset price at time $T$, then the buyer lost money. If the asset ends up being more, then the buyer made money. We want to construct some sort of framework $V$ which solves for the "fair premium" of an option contract with expiration date $T$. Now that these fundamentals are down, let's look at the actual Black-Scholes PDE:

## Delta Hedging

Consider you are a market maker and you sell a call option to a consumer. You owe money if the contract goes up: you are now short an option worth $-V$. Unfortunately, the asset price $S$ moves up and down randomly, meaning you do not know if you are going to end up making or losing money on the contract. So, you construct a portfolio designed to offset this potential loss, if the contract does indeed go up:
$$\Pi=-V+\Delta S$$
In terms of Black-Scholes, $\Delta$ is $\frac{\partial V}{\partial S}$ measures the change in option value per $\$1$ the asset shifts. If the asset shifts $\$1$ and the $\Delta$ is $0.60$, the option becomes $\$60$ more valuable. That means your short position puts you $\$60$ in the red. But, if you buy $0.60$ shares of the stock, you net zero dollars. So, by buying $\Delta$ shares of a stock times the number of option contracts sold, your expected loss/gain is always nothing, meaning the portfolio is completely risk-free. 

This creates the **no-arbitrage** rule, states that the returns of the portfolio must not exceed a risk-free interest rate. Otherwise, this would create severe market asymmetry. 

## Deriving the PDE

In order to derive the Black-Scholes PDE, we must start with the standard Geometric Brownian Motion PDE:
$$dS = \mu S dt + \sigma S dW$$
Where the $\mu$ term is the drift term, and the $\sigma$ term is the stochastic/volatility term. If $V(S,t)$ is the option's value, then we must take the Taylor expansion of the GBM PDE with respect to this. Note that according to Itô's Lemma, $(dt)^2\rightarrow0$, $dtdW\rightarrow0$, and $(dW)^2\rightarrow dt$. Specifically, we must approximate $V(S+dS,t+dt)$ centered around $(S,t)$. This then turns into:
$$dV = \left( \frac{\partial V}{\partial t} + \mu S \frac{\partial V}{\partial S} + \frac{1}{2} \sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} \right) dt + \sigma S \frac{\partial V}{\partial S} dW$$
Now recall our delta hedging formula:
$$\Pi=-V+\Delta S$$
It's derivative is:
$$d\Pi=-dV+\Delta dS$$
If we substitute our taylor expansion into this:
$$d\Pi = -\left[ \left( \frac{\partial V}{\partial t} + \mu S \frac{\partial V}{\partial S} + \frac{1}{2} \sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} \right) dt + \sigma S \frac{\partial V}{\partial S} dW \right] + \Delta (\mu S dt + \sigma S dW)$$
Now we want to try and eliminate any randomness governed by $dW$ inside this equation. Grouping all the terms by randomness yields
$$d\Pi = \left( -\frac{\partial V}{\partial t} - \mu S \frac{\partial V}{\partial S} - \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} + \Delta \mu S \right) dt + \left( \Delta - \frac{\partial V}{\partial S} \right) \sigma S dW$$
We know we want our $\Delta$ to explicitly measure the amount an option's value changes for every $\$1$ the asset price $S$ shifts. Mathematically, this is $\frac{\partial V}{\partial S}$ So, the entire $dW$ term cancels out. And thus, our portfolio is risk-free:
$$d\Pi = \left( -\frac{\partial V}{\partial t} - \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} \right) dt$$
Now we must apply the idea of zero arbitrage. What this means, is that the portfolio grows at a risk free rate $r$:
$$d\Pi = r \Pi dt$$
If we substitute our portfolio definition into this ($\Pi = -V + \frac{\partial V}{\partial S} S$), we get:
$$d\Pi = r \left( -V + \frac{\partial V}{\partial S} S \right) dt$$
Now we can rearrange both $d\Pi$ expressions to be equal to each other:
$$\left( -\frac{\partial V}{\partial t} - \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} \right) dt = r \left( -V + \frac{\partial V}{\partial S} S \right) dt$$
$$-\frac{\partial V}{\partial t} - \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} = -r V + r S \frac{\partial V}{\partial S}$$
$$\frac{\partial V}{\partial t} + \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} + r S \frac{\partial V}{\partial S} - r V = 0$$
Thus, deriving the Black-Scholes PDE.

## Analyzing the PDE

$$\frac{\partial V}{\partial t}+\frac{1}{2}\sigma^2S^2\frac{\partial^2V}{\partial S^2}+rS\frac{\partial V}{\partial S}-rV=0$$

Each term in the PDE corresponds to a "Greek", or a parameter carrying information about an option contract. 
- **Time/Decay $\theta$: $\frac{\partial V}{\partial t}$**, Represents how an option's intrinsic value as time moves forward. Because options have an expiration date, they almost always lose value as time progressives: this value is almost entirely negative for option contracts.
- **Convexity/Gamma $\Gamma$: $\frac 12 \sigma^2 S^2 \frac{\partial^2 V}{\partial S^2}$**, Represents the volatility, or the acceleration of the option value when the asset price shifts. Given that options have a curved payoff profile (expected losses never drop below zero, disregarding the premium), we typically want to maximize curvature, so we usually want a large $\sigma$ value.
- **Expected Drift/Delta $\Delta$: $rS\frac{\partial V}{\partial S}$**, Represents the expected shift of the option value given the asset price pushed by a risk-free interest rate. It directly can measure how much the option value changes for every $\$1$ the asset price shifts.
- **Opportunity Cost/Discounting:** $-rV$, Represents that a contract is inherently a liability or the tying up of capital, growing at a risk-free rate.


## Solving the PDE

Analytically speaking, an asset's value $V(t,S)$ is defined as the expected future payoff under a risk-neutral process. This is known as the Feynman-Kac theorem:
$$
V(t,S)=\mathbb{E}^{\mathbb{Q}}\Big[e^{-r(T-t)}\operatorname{Payoff}(S_T)|S_t=S\Big]
$$
We know that it is relatively impossible to solve this PDE purely using calculus. So, we take the expectation of all possible future payouts, discounted by a risk free rate back to the present day. This effectively maps out all possible paths the PDE can take. 

Now consider how $S_T$ is defined. Assuming it follows a standard random walk, so a Geometric Brownian Motion, a future stock price $S_T$ is defined as:
$$
S_T = S_t\exp\Big(\Big(r-\frac{\sigma^2}{2}\Big)\tau+\sigma\sqrt{\tau}z\Big)
$$
Where $z\sim\mathcal{N}(0,1)$ and $\tau$ is periods into the future. To find the expected value, we have to integrate over a standard bell curve, with mean zero and variance one. The option only pays out when $S_T>K$, so we must find the values of $z$ for when this is true:
$$
S_t\exp\Big(\Big(r-\frac{\sigma^2}{2}\Big)\tau+\sigma\sqrt{\tau}z\Big)>K
$$
We take the natural logarithm of both sides:
$$
\Big(r-\frac{\sigma^2}{2}\Big)\tau+\sigma\sqrt{\tau}z>\ln\Big(\frac{K}{S_T}\Big)
$$
$$
z>\frac{-\ln(K/S_T)-(r-\sigma^2/2)\tau}{\sigma\sqrt{\tau}}
$$
This means that $z>-d_2$. We can then substitute this into our expectation formula in continuous form:
$$
V=e^{-r\tau}\int_{-d_2}^\infty(S_T(z)-K)\frac{1}{\sqrt{2\pi}}e^{\frac{-z^2}{2}}dz
$$
Now we can split up the integral into two pieces: the asset price at time $T$, and the strike price $K$ we would be expected to pay. 
$$V=e^{-r\tau}\int_{d_2}^\infty S_T(z)\phi(z)dz-Ke^{-r\tau}\int_{d_2}^\infty \phi(z)dz$$
The second term is literally just the standard normal distribution from $-d_2$ to $\infty$. This can be rewritten as $N(d_2)$ (because bell curves are symmetrical).

Now let's solve the stock component. Substituting $S_T$ into the formula:
$$
e^{-r\tau}\int_{-d_2}^\infty S_t\exp{\Big(\Big(r-\frac{\sigma^2}{2}\tau+\sigma\sqrt{\tau}z\Big)\Big)}\frac{1}{\sqrt{2\pi}}e^{\frac{-z^2}{2}}dz
$$
The exponential terms simplify to:
$$-r\tau+(r-\frac{\sigma^2}{2})\tau+\sigma\sqrt{\tau}z-\frac{z^2}{2}$$
$$-\frac{\sigma^2}{2}\tau+\sigma\sqrt{\tau}z-\frac{z^2}{2}$$
Through factoring out $\frac 12$, we get a perfect square:
$$
-\frac 12 (z-\sigma\sqrt{\tau})^2
$$
Which means our integral is rewritten as:
$$S_t \int_{-d_2}^{\infty} \frac{1}{\sqrt{2\pi}} \exp\left( -\frac{1}{2}(z - \sigma \sqrt{\tau})^2 \right) dz$$
Now we can use substitution to generalize this integral. If $u=z-\sigma\sqrt{\tau}$, since $z=-d_2$, then  $u=-d_2-\sigma\sqrt{\tau}$. This means $u=-d_1$. This changes the integration bounds to:
$$S_t \int_{-d_1}^{\infty} \frac{1}{\sqrt{2\pi}} e^{-u^2/2} du$$
So, this part evaluates to $S_t N(d_1)$. The entire solution is then:
$$S_tN(d_1)-Ke^{-r\tau}N(d_2)$$


## Black-Scholes Solution

These are simply the dynamics our framework $V$ must adhere to. The solution $V$ is:
$$V(S,t)=C(S,t)=S_0N(d_1)-Ke^{-r(T-t)}N(d_2)$$
Let's look at this solution. $d_1$ is:
$$
d_1=\frac{\ln(S_0/K)+(r+\sigma^2/2)\tau}{\sigma\sqrt{\tau}}
$$ 
Here, $\ln(S_0/K)$ is the log-moneyness, or the condensed ratio of asset price to strike price. $(r+\sigma^2/2)\tau$ is the expected growth of the asset in general. This is all divided by the overall uncertainty of the asset, $\sigma\sqrt{\tau}$, which effectively yields the number of standard deviations the stock price is away from the strike price (this is essentially a modified z-score formula). Fitting this into the normal distribution $N$ yields the actual probability, which we discount $S_0$ by in the actual solution: a greater probability means a higher chance of profits because it corresponds to a higher z-score.

$d_2$ is:
$$d_2=d_1-\sigma\sqrt{\tau}$$
Given that $d_1$ is the number of standard deviations the asset price is away from the strike price, subtracting the total volatility $\sigma\sqrt{\tau}$ is a more conservative score. This is necessary for mimicking real-world financial dynamics. $N(d_2)$ is effectively the probability an option contract expires "in-the-money", or when the asset price is greater than the strike price (meaning the contract is excercised). We discount $Ke^{-r\tau}$ by this probability to account for how likely we are to actually pay the strike price. And so, the Black-Scholes solution is effectively:
$$
\operatorname{Expected Asset Value at }T - \operatorname{Expected Amount to Pay at }T
$$
Thus giving the "fair premium" of the contract. Say we calculate this fair premium to be $C(S,t)=\$2.50$, and it's being sold at $\$4.00$. So, we know the contract is overpriced and we wouldn't want to purchase it. 

## .In, At, and Out of the Money

Let's define what ITM, ATM, and OTM volatilities are. These are all "implied" volatilities, meaning they refer to the volatility given specific characteristics of an option contract. Note that an option contract an be either in-the-money, at-the-money, or out-of-the-money.

- An ITM contract is when the asset price is greater than the strike price ($S_T>K$) for a European call option, and when the asset price is lesser than the strike price ($S_T<K$) for a European put option. You stand to make money in both scenarios. 
- An ATM contract is when the asset price is equal to the strike price in either a European call or put option ($S_T=K$). You stand to net nothing in either scenario. ATM volatility reveals the expected volatility of contracts of revolving around a specific asset, setting an anchor point for market behavior. 
- An OTM contract with the converse of an ITM contract, when the strike price is greater than the asset price for a call option ($S_T<K$), or when the asset price is greater than the strike price for a put ($S_T>K$). You stand to lose money in either scenario. OTM contracts can actually act as "crash insurance": an OTM put can offset losses in a portfolio, since it doesn't necessarily have to be cashed. A higher OTM volatility compared to ATM volatility reveals a greater general fear of a market crash.

Now let's define the Newton-Raphson method, which is used to actually determine implied volatility. The formula for this is follow:
$$
\sigma_{new} = \sigma_{old} - \frac{C_{model} - C_{market}}{\text{Vega}}
$$
$C_{market}$ refers to the current premium that the market is charging for the specific contract. $C_{model}$ is our sandbox. We may start with an initial volatility guess, typically around $\sigma=0.80$, and then solve $C_{model}=C(\sigma=0.80)$. In this scenario, we can also alter our strike $K$ accordingly. If we want to solve for ATM volatility, set the strike equal to the asset price. If we are gauging the volatility of an OTM put, set the strike to maybe $0.90S$.

$\text{Vega}$ represents how much the value of the option shifts given it's volatility. Mathematically, it's defined as $\frac{\partial C}{\partial \sigma}$. It's formula is:
$$\operatorname{Vega}=S_0 \cdot \sqrt{\tau} \cdot N'(d_1)$$
Since $N'(d_1)$ is greatest when $d_1$ is close to zero, this proves that ATM contracts are the riskiest. This increases vega, decreasing the volatility differential. Increasing overall volatility. And so, this should effectively yield a more realistic volatility, which serves as an anchor point for options contracts revolving around a given asset. 

But, we cannot use the Newton-Raphson for ITM call options. This is because they inherently have a high standard deviation, which shrinks vega down to zero. In order to get around this, we use put-call parity, which states that the IV for an ITM call is the same for an OTM put. In that case, we just use:
$$\sigma_{\text{new}} = \sigma_{\text{old}} - \frac{P_{\text{BS}}(\sigma_{\text{old}}) - P_{\text{market}}}{\text{Vega}(\sigma_{\text{old}})}$$
where 
$$P_{\text{BS}}(\sigma) = K e^{-rt} N(-d_2) - S_0 N(-d_1)$$
And $P_{\text{market}}$ is the price for an OTM put of the same asset at the same strike price.
