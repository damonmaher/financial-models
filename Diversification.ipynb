# Portfolio Diversification

## Mean Variance Optimization

This is the foundation for modern portfolio theory. It specifically provides a framework which maximizes the expected returns of a portfolio, while minimizing risk. There are three statistical pillars which make up this framework:
- **Expected Return ($\boldsymbol{\mu}$):** The average return of a portfolio, considering how significant an asset is inside the portfolio:
$$\operatorname{Inner Product Matrix:~}E(R_p)=w^T\mu=\sum_{i=1}^nw_i\mu_i$$
- **Variance/Risk ($\boldsymbol{\sigma^2}$):** The measure of how much a portfolio's returns deviate from the mean.
- **Covariance ($\Sigma$):** The measure of how assets move together:
$$\sigma^2_p=w^T\Sigma w=\sum_{i=1}^n\sum_{j=1}^nw_iw_j\sigma_{ij}$$

### Derivation

The goal of MVO is to find weights which minimize portfolio variance for a specific target return:
$$\min_w\frac 12w^T\Sigma w$$
But note that we must enforce two constraints: the portfolio must reach the target return ($w^T\mu = \mu_p$), and all weights must sum to $1$. This is specifically solved using Lagrangian multipliers. This solves for the minimum of a function considering a constraint $g(x,y,z)=c$:
$$\mathcal{L}(x,y,z,\lambda)=f(x,y,z)-\lambda(g(x,y,z)-c)$$
If our constraints are expressed as $w^T\mu = \mu_p$ and $w^T\boldsymbol{1}=1$, then our equation in this case is:
$$\mathcal{L}(w, \lambda_1, \lambda_2)=\frac 12w^T\Sigma w-\lambda_1(w^T\mu-\mu_p)-\lambda_2(w^T\mathbf{1}-1)$$
We can solve for $w$ by taking the derivative with respect to $w^T$:
$$\frac{\partial L}{\partial w^T}=\Sigma w-\lambda_1\mu-\lambda_2\mathbf{}1=0$$
$$w=\Sigma^{-1}(\lambda_1\mu+\lambda_2\mathbf{1})$$
Now, we can solve for $\lambda_1$ and $\lambda_2$. Consider the transpose of the weight vector:
$$w^T=\lambda_1\mu^T\Sigma^{-1}+\lambda_2\mathbf{1}^T\Sigma^{-1}$$
Now we have to create "Merton's constants". These are necessary in order to easily set up a nice system in order to solve for our two $\lambda$ values:
- $A=\mathbf{1^T}\Sigma^{-1}\mathbf{1}$
- $B=\mu^T\Sigma^{-1}1=1^T\Sigma^{-1}\mu$ 
- $C=\mu^T\Sigma^{-1}\mu$
Consider our first constraint $w^T\mu = \mu_p$:
$$(\lambda_1\mu^T\Sigma^{-1}+\lambda_2\mathbf{1}^T\Sigma^{-1})\mu=\mu_p$$
$$\lambda_1C+\lambda_2B=\mu_p$$
And the second constraint $w^T\boldsymbol{1}=1$
$$(\lambda_1\mu^T\Sigma^{-1}+\lambda_2\mathbf{1}^T\Sigma^{-1})\mathbf{1}=1$$
$$\lambda_1B+\lambda_2A=1$$
This yields us a $2\times2$ matrix:
$$
\begin{bmatrix}
B&A\\
C&B
\end{bmatrix}
\begin{bmatrix}
\lambda_2\\
\lambda_1
\end{bmatrix}=
\begin{bmatrix}
1\\
\mu_p
\end{bmatrix}
$$
Rearranging to solve for our two lambda values leaves us with:
$$\begin{bmatrix} \lambda_1 \\ \lambda_2 \end{bmatrix} = \frac{1}{AC-B^2} \begin{bmatrix} A & -B \\ -B & C \end{bmatrix} \begin{bmatrix} \mu_p \\ 1 \end{bmatrix}$$
$$\lambda_1 = \frac{A \mu_p - B}{D}$$$$\lambda_2 = \frac{C - B \mu_p}{D}$$
Given that we now know $\lambda_1$ and $\lambda_2$, we can now solve for our optimal weight vector using $w=\Sigma^{-1}(\lambda_1\mu+\lambda_2\mathbf{1})$.

## MVO's Shortfalls

While MVO is mathematically elegant, in practice it is actually referred to as an "error maximizer". MVO is incredibly sensitive to return estimates, so tweaking a single asset's return by as low as $0.2%$ can do so much as reallocate $40%$ of the portfolio; MVO also does not omit the heteroskedasticity of returns, and thus factors in noise completely; finally, MVO has no benchmark and cannot assume prior market equilibrium. And so, this process creates ridiculous portfolio allocations.

This is where the Black-Litterman model comes in. It specifically combines **market equilibrium theory** with modern **Bayesian statistics**. It calculates the the **implied equilibrium returns $\boldsymbol{\Pi}$** that would theoretically cause a rational market to hold the current market capitalization weights:
$$\Pi = \lambda \Sigma w_{mkt}$$
Here $\lambda$ is a risk-aversion parameter and $\Sigma$ is the covariance matrix. If a portfolio manager doens't have unique views, then the current market portoflio is the default neutral starting point. 

Investors will also specify their explicit views (*"Asset A will outperform asset B by 2%"*) and assign a matrix of confidence levels $(\Omega)$ representing their certainty. Then, they will use **Bayesian blending** to combine the market equilibrium baseline $\Pi$ with a portfolio manager's views $Q$ using Bayes' Theorem to generate a new posterior expected returns vector 
$$\mu_{BL} = \left[(\tau \Sigma)^{-1} + P^T \Omega^{-1} P\right]^{-1} \left[(\tau \Sigma)^{-1} \Pi + P^T \Omega^{-1} Q\right]$$
Plugging this value into MVO generates much more well-diversified and stable portfolio allocations. 

## Analyzing the Pipeline

### Assumptions

Let's walk through an example of calculating $\mu_{BL}$. Let's consider our initial market capitalization weights $w_{mkt}$ to be $w_{mkt} = \begin{bmatrix} 0.60 \\ 0.40 \end{bmatrix}$ (*assuming 60% US market cap, 40% international*). Also, let's assume the following: $\sigma_1=15%$, $\sigma_2=20%$, and $\rho_{12}=0.50$. That means $\operatorname{Covar}(1,2)=0.50\times0.15\times0.20=0.0150$. Our covariance matrix is then:
$$\Sigma = \begin{bmatrix} 0.0225 & 0.0150 \\ 0.0150 & 0.0400 \end{bmatrix}$$

### Risk Aversion Coefficient ($\lambda$)

$\lambda = \frac{E(R_{mkt}) - R_f}{\sigma_{mkt}^2}$, where $E(R_{mkt})-R_f$ is the expected equity risk premium (expected returns minus risk-free rate). Typically $\lambda\approx2.5$.

### Implied Equilibrium Returns ($\Pi$)

Implied equilibrium returns are $\Pi = \lambda \Sigma w_{mkt}$:
$$\Pi = 2.5 \begin{bmatrix} 0.0225 & 0.0150 \\ 0.0150 & 0.0400 \end{bmatrix} \begin{bmatrix} 0.60 \\ 0.40 \end{bmatrix} = 2.5 \begin{bmatrix} 0.0195 \\ 0.0250 \end{bmatrix} = \begin{bmatrix} 0.04875 \\ 0.06250 \end{bmatrix} = \begin{bmatrix} 4.875\% \\ 6.250\% \end{bmatrix}$$
It specifically describes the returns the market must expect in order for the market capital allocation to be optimal. 

Note that $\tau$ is a scaling factor that describes how uncertain we are about the expected equilibrium returns relative to the variance of returns. It is typically either $0.02$ or $0.05$. 

### Investor Views Vector ($Q$) & Pick Matrix ($P$)

Consider the view that *"U.S. Equities will return 7.00%"*. Then, $Q = \begin{bmatrix} 0.07 \end{bmatrix}$. The pick matrix $P = \begin{bmatrix} 1 & 0 \end{bmatrix}$ then maps this belief to asset 1.  

### View Uncertainty Matrix ($\Omega$)

This quantifies the variance of the manager's view. It's calculated via $\Omega = P (\tau \Sigma) P^T$:
$$\tau \Sigma = 0.05 \begin{bmatrix} 0.0225 & 0.0150 \\ 0.0150 & 0.0400 \end{bmatrix} = \begin{bmatrix} 0.001125 & 0.000750 \\ 0.000750 & 0.002000 \end{bmatrix}$$$$\Omega = \begin{bmatrix} 1 & 0 \end{bmatrix} \begin{bmatrix} 0.001125 & 0.000750 \\ 0.000750 & 0.002000 \end{bmatrix} \begin{bmatrix} 1 \\ 0 \end{bmatrix} = \begin{bmatrix} 0.001125 \end{bmatrix}$$
As you can see, this matrix multiplication yields a higher result if there is a greater uncertainty $\tau$ and a greater covariance matrix $\Sigma$ yield a greater uncertainty view $\Omega$. Once all of these are found, we use the above formula to solve for $\mu_{BL}$.
